2.  Complete the first bullet of the Check-in before Section 7.3.

    Write a function `sqrt_pos_unvec()` that takes in a single value as an argument, and returns the square root if the value is positive. Then, write a for loop that uses this function to construct a new vector where the positive values are square rooted.

```python
import numpy as np

def sqrt_pos_unvec(x):
    """Return the square root of x if x is positive; otherwise return x unchanged."""
    if x > 0:
        return np.sqrt(x)
    return x

test_vec = np.array([-4, 9, -1, 16, 0, 25])

result = []
for val in test_vec:
    result.append(sqrt_pos_unvec(val))

np.array(result)
```

3.  Complete the second bullet of the Check-in before Section 7.3.

    Write a function `sqrt_pos_vec()` that takes in a vector of values, and returns a vector with the positive values square rooted. Do not use a for loop inside your function.

```python
def sqrt_pos_vec(x):
    """Square root the positive values of a vector, leave the rest unchanged (no loop)."""
    x = np.asarray(x, dtype=float)
    return np.where(x > 0, np.sqrt(np.abs(x)), x)

sqrt_pos_vec(test_vec)
```

4.  Complete the Check-in before Section 7.4.
    - a. Write a `sing_verse_3()` function that also lets us change the container (e.g. bottle, can, …) at each step of the loop.
    - b. Use `map()` to sing a few verses.
    - c. What happens if you supply three different drinks, but only two different types of containers?

```python
# a. sing_verse_3() lets both the drink and the container vary
def sing_verse_3(num, drink, container):
    # pluralize the container ("bottle" -> "bottles", "glass" -> "glasses")
    container = container + ("es" if container.endswith(("s", "sh", "ch")) else "s")
    return (f"{num} {container} of {drink} on the wall, {num} {container} of {drink}.\n"
            f"Take one down, pass it around, {num - 1} {container} of {drink} on the wall.\n")

# b. Use map() to sing a few verses
nums = [99, 98, 97]
drinks = ["soda", "juice", "tea"]
containers = ["bottle", "can", "cup"]

verses = list(map(sing_verse_3, nums, drinks, containers))
for v in verses:
    print(v)

# c. Three drinks but only two containers
short_containers = ["bottle", "can"]
verses_c = list(map(sing_verse_3, nums, drinks, short_containers))
# map() stops at the SHORTEST input, so only 2 verses are sung:
# the third drink ("tea") is silently dropped -- no error is raised.
print(len(verses_c))
for v in verses_c:
    print(v)
```

5.  Complete the Check-in before Section 7.5.

    Use a lambda function with `sing_verse_3()` to sing a few verses about milk in glasses.

```python
# Lambda fixes drink = "milk" and container = "glass"; only the number varies
milk_verses = list(map(lambda n: sing_verse_3(n, "milk", "glass"), range(5, 0, -1)))
for v in milk_verses:
    print(v)
```

6. Load the `penguins` dataset from the `palmerpenguins` library, as well as any other libraries you need. Then write a function that takes in information about a penguin, and returns one of the following definitions:

*   "Big Mouth Billy": Male penguins with bill length times bill depth greater than 800.
*   "Dainty Daisy": Female penguins with flipper length less than 5% of body mass.
*   "Average Adelie:" Any Adelie penguins that do not fall into either category.
*   "Other": Any penguins that do not fall into any of the categories.



```python
import pandas as pd
from palmerpenguins import load_penguins

penguins = load_penguins()

def penguin_category(species, sex, bill_length, bill_depth, flipper_length, body_mass):
    """Classify a penguin into one of four categories."""
    if sex == "male" and bill_length * bill_depth > 800:
        return "Big Mouth Billy"
    elif sex == "female" and flipper_length < 0.05 * body_mass:
        return "Dainty Daisy"
    elif species == "Adelie":
        return "Average Adelie"
    else:
        return "Other"

penguins.head()
```

7. Use an iterable function to create a new variable in the data frame called `category_name` that adds these labels.

```python
penguins["category_name"] = list(map(
    penguin_category,
    penguins["species"],
    penguins["sex"],
    penguins["bill_length_mm"],
    penguins["bill_depth_mm"],
    penguins["flipper_length_mm"],
    penguins["body_mass_g"],
))

penguins[["species", "sex", "bill_length_mm", "bill_depth_mm",
          "flipper_length_mm", "body_mass_g", "category_name"]].head(10)
```
