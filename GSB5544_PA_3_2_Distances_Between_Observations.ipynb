## Ames - Recommending Similar Homes

1\. Suppose that you really like house 0 in the Ames housing data set, but it is too expensive. Find cheaper homes that are similar to it --- in terms of living area, number of bedrooms, number of bathrooms --- by calculating distances from house 0. You might want to try different distance metrics and different scaling methods; how sensitive are your results to these choices?

Be sure to actually look at the profiles of the homes that your algorithm picked out as most similar (based on these 3 variables). Do they make sense?

_Think:_ If the goal is to find a "good deal" on a similar house, should sale price be included as a variable in your distance metric?

```python
df_ames = pd.read_csv("https://raw.githubusercontent.com/kevindavisross/data301/main/data/AmesHousing.txt", sep="\t")
```

```python
# YOUR CODE HERE. ADD CELLS AS NEEDED
df_ames["Bathrooms"] = df_ames["Full Bath"] + df_ames["Bsmt Full Bath"]

house0 = df_ames.loc[0]

show_vars = ["Gr Liv Area", "Bedroom AbvGr", "Bathrooms", "House Style", "Year Built", "SalePrice"]

house0[show_vars]
```

```python
features = ["Gr Liv Area", "Bedroom AbvGr", "Bathrooms"]
X = df_ames[features]. astype(float)

X_z = (X - X.mean()) / X.std()

diff = X_z -X_z.loc[0]

df_ames["dist_euclid"] = np.sqrt((diff) **2).sum(1)
df_ames["dist_manhattan"] = diff.abs().sum(axis =1)
```

```python
df_cheaper_sales_price = df_ames[df_ames["SalePrice"]<house0["SalePrice"]]

df_cheaper_sales_price.sort_values("dist_euclid")[show_vars]
```

```python
df_cheaper_sales_price.sort_values("dist_euclid")[show_vars].head(10)
```

```python
compare_vars = ["Gr Liv Area", "Bedroom AbvGr", "Bathrooms"]

df_scaled = (
    df_ames[compare_vars] - df_ames[compare_vars].mean()
) / df_ames[compare_vars].std()
```

```python
df_ames["dist_scaled"] = np.sqrt(
    ((df_scaled - df_scaled.loc[0]) ** 2).sum(axis=1)
)
```

```python
df_cheaper_sales_price = df_ames[
    df_ames["SalePrice"] < house0["SalePrice"]
]
```

```python
show_vars_scaled = [
    "SalePrice",
    "Gr Liv Area",
    "Bedroom AbvGr",
    "Bathrooms",
    "dist_scaled"
]

df_cheaper_sales_price.sort_values("dist_scaled")[show_vars_scaled].head(10)
```

```python
compare_vars = ["Gr Liv Area", "Bedroom AbvGr", "Bathrooms"]

df_scaled = (
    df_ames[compare_vars] - df_ames[compare_vars].mean()
) / df_ames[compare_vars].std()
```

```python
df_ames["dist_scaled"] = np.sqrt(
    ((df_scaled - df_scaled.loc[0]) ** 2).sum(axis=1)
)
```

```python
df_cheaper_sales_price = df_ames[
    df_ames["SalePrice"] < house0["SalePrice"]
]

show_vars_scaled = [
    "SalePrice",
    "Gr Liv Area",
    "Bedroom AbvGr",
    "Bathrooms",
    "dist_scaled"
]

df_cheaper_sales_price.sort_values("dist_scaled")[show_vars_scaled].head(10)
```

**YOUR RESPONSE HERE**

The similar homes make sense because they have smiliar living around 1,652-1660sq.ft, 3 bedrooms, and similiar bathrooms. Scaling did not change the results much. Sale price should not be included because we want to find similar homes that are cheaper

## Colleges similar to Cal Poly

We'll use data from the [College Scorecard data](https://collegescorecard.ed.gov/) to find colleges and universities that are similar to Cal Poly.

```python
df_college = pd.read_csv("https://datasci112.stanford.edu/data/college_attributes.csv")

df_college.set_index("Institution", inplace = True)

df_college
```

We'll want to single out Cal Poly, which we can do like this.

```python
school_name = "California Polytechnic State University-San Luis Obispo"

cp = df_college.loc[school_name]

cp
```

1\. Based on only the admission rate and the number of undergraduates, what schools are most similar to Cal Poly? Specify how you're making this decision.

```python
# YOUR CODE HERE. ADD CELLS AS NEEDED
df_college.columns

```

```python
compare_vars = ["AdmissionRate", "Undergraduates"]

df_college[compare_vars] = df_college[compare_vars].apply(
    pd.to_numeric, errors="coerce"
)
```

```python
college_scaled = (
    df_college[compare_vars] - df_college[compare_vars].mean()
) / df_college[compare_vars].std()

cp_scaled = college_scaled.loc[school_name]
```

```python
df_college["distance"] = np.sqrt(
    ((college_scaled - cp_scaled) ** 2).sum(axis=1)
)
```

```python
df_college.sort_values("distance")[
    ["AdmissionRate", "Undergraduates", "distance"]
].head(11)
```

**YOUR RESPONSE HERE**

I standardized admission rate and number of undergraduates, then used Euclidean distance to measure similarity to Cal Poly. The schools with the smallest distances were considered most similar. The closest match had a distance of 0.3092, meaning its admission rate and undergraduate enrollment were most similar to Cal Poly among the other schools.

Now consider the admission rate, the number of undergraduates, and also the [Carnegie classification](https://en.wikipedia.org/wiki/Carnegie_Classification_of_Institutions_of_Higher_Education) of the type of school, and the ownership (public, private, etc.) Based on these variables, what schools are most similar to Cal Poly?
```python
cp_scaled2 = college_scaled2.loc[school_name]
```

```python
df_college2.sort_values("distance2")[
    [
        "AdmissionRate",
        "Undergraduates",
        "CarnegieClassification",
        "Ownership",
        "distance2"
    ]
].head(11)
```

**YOUR RESPONSE HERE**

I standardized the quantitative variables and used Euclidean distance, while also including Carnegie classification and ownership. The closest schools are public and have the same Carnegie classification as Cal Poly. Adding these categorical variables changed the matches and made them more similar in overall school type.
